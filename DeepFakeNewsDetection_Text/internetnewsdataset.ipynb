```python
import pandas as pd

df_business = pd.read_csv("/kaggle/input/datasets/palakjaiswal24/internetnewsdatasetfinal/archive (13)/business_data.csv")
df_education = pd.read_csv("/kaggle/input/datasets/palakjaiswal24/internetnewsdatasetfinal/archive (13)/education_data.csv")
df_entertainment = pd.read_csv("/kaggle/input/datasets/palakjaiswal24/internetnewsdatasetfinal/archive (13)/entertainment_data.csv")
df_sports = pd.read_csv("/kaggle/input/datasets/palakjaiswal24/internetnewsdatasetfinal/archive (13)/sports_data.csv")
df_technology = pd.read_csv("/kaggle/input/datasets/palakjaiswal24/internetnewsdatasetfinal/archive (13)/technology_data.csv")

print("Business:", len(df_business))
print("Education:", len(df_education))
print("Entertainment:", len(df_entertainment))
print("Sports:", len(df_sports))
print("Technology:", len(df_technology))
```

```python
print(df_business.head())
print(df_business.columns)
```

```python
def standardize_news(df, source_name):
    df["text"] = (
        df["headlines"].fillna('') + " " +
        df["description"].fillna('') + " " +
        df["content"].fillna('')
    )
    
    df_clean = pd.DataFrame({
        "text": df["text"],
        "label": 0,
        "language": "en",
        "source": source_name
    })
    
    df_clean = df_clean[df_clean["text"].str.strip() != ""]
    
    return df_clean
```

```python
df_business_clean = standardize_news(df_business, "india_business")
df_education_clean = standardize_news(df_education, "india_education")
df_entertainment_clean = standardize_news(df_entertainment, "india_entertainment")
df_sports_clean = standardize_news(df_sports, "india_sports")
df_technology_clean = standardize_news(df_technology, "india_technology")
```

```python
df_classification = pd.concat([
    df_business_clean,
    df_education_clean,
    df_entertainment_clean,
    df_sports_clean,
    df_technology_clean
], ignore_index=True)

print("Total merged:", len(df_classification))
```

```python
df_classification["length"] = df_classification["text"].apply(lambda x: len(str(x).split()))

print(df_classification["length"].describe())
```

```python
df_classification = df_classification[df_classification["length"] > 30]
df_classification = df_classification[df_classification["length"] < 6000]
```

```python
df_classification = df_classification.drop_duplicates(subset=["text"])

print("After deduplication:", len(df_classification))
```

```python
df_classification.to_csv("india_news_classification_cleaned.csv", index=False)

print("Saved successfully.")
print("Final count:", len(df_classification))
print(df_classification["label"].value_counts())
```

```python
import pandas as pd

df_india_headlines = pd.read_csv("india_headlines_2001_2023_cleaned.csv")
df_indian_recent   = pd.read_csv("indian_recent_news_cleaned.csv")
df_inshorts_en     = pd.read_csv("inshorts_cleaned.csv")
df_classification  = pd.read_csv("india_news_classification_cleaned.csv")
df_toi             = pd.read_csv("times_of_india_cleaned.csv")
df_bbc_en          = pd.read_csv("bbc_2022_2024_cleaned.csv")
df_huff            = pd.read_csv("huffpost_cleaned.csv")

df_bbc_hi          = pd.read_csv("bbc_hindi_cleaned.csv")
df_inshorts_hi     = pd.read_csv("inshorts_hindi_cleaned.csv")
```

```python
samples = {
    "india_headlines": 6000,
    "indian_recent": 4000,
    "inshorts_en": 6000,
    "classification": 6000,
    "toi": 4000,
    "bbc_en": 2800,
    "huff": 1200,
    "bbc_hi": 2000,
    "inshorts_hi": 8000
}

df_new_real = pd.concat([
    df_india_headlines.sample(samples["india_headlines"], random_state=42),
    df_indian_recent.sample(samples["indian_recent"], random_state=42, replace=True),
    df_inshorts_en.sample(samples["inshorts_en"], random_state=42),
    df_classification.sample(samples["classification"], random_state=42),
    df_toi.sample(samples["toi"], random_state=42),
    df_bbc_en.sample(samples["bbc_en"], random_state=42),
    df_huff.sample(samples["huff"], random_state=42),
    df_bbc_hi.sample(samples["bbc_hi"], random_state=42),
    df_inshorts_hi.sample(samples["inshorts_hi"], random_state=42)
], ignore_index=True)
```

```python
import pandas as pd

liar_train = pd.read_csv("/kaggle/input/datasets/palakjaiswal24/liardatasetmerge/train.tsv", sep="\t", header=None)
liar_valid = pd.read_csv("/kaggle/input/datasets/palakjaiswal24/liardatasetmerge/valid.tsv", sep="\t", header=None)
liar_test  = pd.read_csv("/kaggle/input/datasets/palakjaiswal24/liardatasetmerge/test.tsv",  sep="\t", header=None)

liar_df = pd.concat([liar_train, liar_valid, liar_test], ignore_index=True)

print("Total LIAR samples:", len(liar_df))
```

```python
liar_df = liar_df[[1, 2]]
liar_df.columns = ["label_raw", "text"]

print(liar_df["label_raw"].value_counts())
```

```python
fake_labels = ["pants-fire", "false", "barely-true"]
real_labels = ["mostly-true", "true"]

liar_df = liar_df[liar_df["label_raw"] != "half-true"]

liar_df["label"] = liar_df["label_raw"].apply(
    lambda x: 1 if x in fake_labels else 0
)

liar_df = liar_df[liar_df["label_raw"].isin(fake_labels + real_labels)]

print(liar_df["label"].value_counts())
```

```python
liar_df = liar_df[liar_df["text"].str.strip() != ""]

liar_df["length"] = liar_df["text"].apply(lambda x: len(str(x).split()))

print(liar_df["length"].describe())
```

```python
liar_df = liar_df[liar_df["length"] > 3]
liar_df = liar_df[liar_df["length"] < 100]
```

```python
liar_fake = liar_df[liar_df["label"] == 1].sample(3000, random_state=42)
liar_real = liar_df[liar_df["label"] == 0].sample(3000, random_state=42)

liar_balanced = pd.concat([liar_fake, liar_real], ignore_index=True)

liar_balanced["language"] = "en"
liar_balanced["source"] = "liar_claims"

print("Final LIAR injection size:", len(liar_balanced))
print(liar_balanced["label"].value_counts())
```

```python
liar_balanced.to_csv("liar_balanced.csv", index=False)

print("liar_balanced.csv saved successfully.")
```

```python
import pandas as pd

# OLD TRAIN
df_old = pd.read_csv("/kaggle/input/datasets/palakjaiswal24/myolddataset/train_final.csv")

# 40k new real layer
df_new_real_40k = pd.read_csv("/kaggle/working/new_real_layer_40k.csv")

# LIAR balanced (6000)
df_liar = pd.read_csv("/kaggle/working/liar_balanced.csv")  # if already saved
```

```python
TARGET_NEW_REAL = 18000

# Calculate sampling ratio
ratio = TARGET_NEW_REAL / len(df_new_real_40k)

print("Sampling ratio:", ratio)

df_new_real_18k = (
    df_new_real_40k
    .groupby("source", group_keys=False)
    .apply(lambda x: x.sample(int(len(x) * ratio), random_state=42))
)

print("New Real Sampled Size:", len(df_new_real_18k))
print(df_new_real_18k["source"].value_counts())
```

```python
remaining = TARGET_NEW_REAL - len(df_new_real_18k)

if remaining > 0:
    extra = df_new_real_40k.sample(remaining, random_state=42)
    df_new_real_18k = pd.concat([df_new_real_18k, extra], ignore_index=True)

print("Final 18k size:", len(df_new_real_18k))
```

```python
print(df_liar["label"].value_counts())
```

```python
df_final = pd.concat([
    df_old,
    df_new_real_18k,
    df_liar
], ignore_index=True)
```

```python
df_final = df_final.sample(frac=1, random_state=42).reset_index(drop=True)

print("Final Dataset Size:", len(df_final))
print(df_final["label"].value_counts())
```

```python
df_final.to_csv("final_training_dataset_refined.csv", index=False)

print("Saved successfully.")
```
