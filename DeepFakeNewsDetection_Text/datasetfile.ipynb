```python
import os
import pandas as pd

base_path = "/kaggle/input/datasets/palakjaiswal24/gujarat"  # adjust folder name

fake_path = os.path.join(base_path, "Gujarati_fake_news")
real_path = os.path.join(base_path, "Gujarati_real_news")

data = []

# Fake files
for filename in os.listdir(fake_path):
    with open(os.path.join(fake_path, filename), 'r', encoding='utf-8', errors='ignore') as f:
        text = f.read().strip()
        if len(text) > 0:
            data.append({
                "text": text,
                "label": 1,
                "language": "gu",
                "source": "regional_gujarati"
            })

# Real files
for filename in os.listdir(real_path):
    with open(os.path.join(real_path, filename), 'r', encoding='utf-8', errors='ignore') as f:
        text = f.read().strip()
        if len(text) > 0:
            data.append({
                "text": text,
                "label": 0,
                "language": "gu",
                "source": "regional_gujarati"
            })

df_gu = pd.DataFrame(data)

print("Total:", len(df_gu))
print(df_gu["label"].value_counts())

df_gu.to_csv("regional_gujarati_cleaned.csv", index=False)
```

```python
# Balanced sampling: 3000 per class
df_gu_fake = df_gu[df_gu["label"] == 1].sample(n=3000, random_state=42)
df_gu_real = df_gu[df_gu["label"] == 0].sample(n=3000, random_state=42)

df_gu_sampled = pd.concat([df_gu_fake, df_gu_real], ignore_index=True)

print("Sampled Total:", len(df_gu_sampled))
print(df_gu_sampled["label"].value_counts())

df_gu_sampled.to_csv("regional_gujarati_cleaned.csv", index=False)
```

COMBINE DATASETS

```python
import pandas as pd

files = [
    "wel_fake_cleaned.csv",
    "fake45k_cleaned.csv",
    "eng_hindi_en_cleaned.csv",
    "eng_hindi_hi_cleaned.csv",
    "hindi_extra_cleaned.csv",
    "indian_fake_real_cleaned.csv",
    "regional_gujarati_cleaned.csv",
    "regional_marathi_cleaned.csv",
    "regional_telugu_cleaned.csv",
    "bengali_cleaned.csv",
    "isot_cleaned.csv"
]

dfs = [pd.read_csv(f) for f in files]

df_all = pd.concat(dfs, ignore_index=True)

print("Total merged:", len(df_all))
print(df_all["label"].value_counts())
print(df_all["language"].value_counts())
```

```python
df_all = df_all.dropna(subset=["text"])
df_all = df_all[df_all["text"].str.strip() != ""]

print("After removing empty:", len(df_all))
```

```python
df_all["length"] = df_all["text"].apply(lambda x: len(str(x).split()))

print(df_all["length"].describe())

df_all = df_all[(df_all["length"] > 100) & (df_all["length"] < 8000)]

print("After length filtering:", len(df_all))
```

```python
df_all = df_all.drop_duplicates(subset=["text"])

print("After deduplication:", len(df_all))
```

```python
df_all = df_all.sample(frac=1, random_state=42).reset_index(drop=True)

print("Final dataset size:", len(df_all))
print(df_all["label"].value_counts())
print(df_all["language"].value_counts())
```

```python
# Separate English and others
df_en = df_all[df_all["language"] == "en"]
df_non_en = df_all[df_all["language"] != "en"]

# Sample 70k English
df_en_sampled = df_en.sample(n=70000, random_state=42)

# Merge back
df_balanced = pd.concat([df_en_sampled, df_non_en], ignore_index=True)

# Shuffle
df_balanced = df_balanced.sample(frac=1, random_state=42).reset_index(drop=True)

print("Final balanced size:", len(df_balanced))
print(df_balanced["language"].value_counts())
print(df_balanced["label"].value_counts())
```

```python
print(df_balanced.groupby("language")["label"].value_counts())
```

```python
# Separate Marathi
df_mr = df_balanced[df_balanced["language"] == "mr"]

# Sample equal real count
min_count = min(df_mr["label"].value_counts())

df_mr_fake = df_mr[df_mr["label"] == 1].sample(n=min_count, random_state=42)
df_mr_real = df_mr[df_mr["label"] == 0].sample(n=min_count, random_state=42)

df_mr_balanced = pd.concat([df_mr_fake, df_mr_real])

# Remove old Marathi
df_no_mr = df_balanced[df_balanced["language"] != "mr"]

# Add balanced Marathi
df_balanced = pd.concat([df_no_mr, df_mr_balanced]).reset_index(drop=True)

print(df_balanced.groupby("language")["label"].value_counts())
```

```python
from sklearn.model_selection import train_test_split

df_balanced["stratify_col"] = (
    df_balanced["language"] + "_" + df_balanced["label"].astype(str)
)

train_df, temp_df = train_test_split(
    df_balanced,
    test_size=0.2,
    stratify=df_balanced["stratify_col"],
    random_state=42
)

val_df, test_df = train_test_split(
    temp_df,
    test_size=0.5,
    stratify=temp_df["stratify_col"],
    random_state=42
)

print("Train:", len(train_df))
print("Val:", len(val_df))
print("Test:", len(test_df))
```

```python
train_df = train_df.drop(columns=["stratify_col"], errors="ignore")
val_df = val_df.drop(columns=["stratify_col"], errors="ignore")
test_df = test_df.drop(columns=["stratify_col"], errors="ignore")
```
